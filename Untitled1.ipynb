```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             classification_report, roc_auc_score, confusion_matrix)
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

tf.random.set_seed(42)
np.random.seed(42)
rng = np.random.default_rng(42)

IMG_SIZE = (64, 64)
TRAIN_PER_CLASS = 600     # images per class (train + validation)
TEST_PER_CLASS = 150      # images per class (test)
EPOCHS = 40

# ------------------------------------------------------------------
# 1. Dataset (NO internet needed)
#    - If you have real photos in  data/cats_and_dogs_filtered/{train,validation}/{cats,dogs}
#      the script uses them.
#    - Otherwise it generates cat-like and dog-like images by itself,
#      so the code ALWAYS runs without any download or error.
# ------------------------------------------------------------------
REAL_ROOT = os.path.join("data", "cats_and_dogs_filtered")


def load_real(split, limit):
    X, y = [], []
    for label, name in enumerate(["cats", "dogs"]):
        folder = os.path.join(REAL_ROOT, split, name)
        files = [f for f in sorted(os.listdir(folder))
                 if f.lower().endswith((".jpg", ".jpeg", ".png"))][:limit]
        for f in files:
            img = Image.open(os.path.join(folder, f)).convert("RGB").resize(IMG_SIZE)
            X.append(np.array(img, dtype="uint8"))
            y.append(label)
    return np.array(X, dtype="uint8"), np.array(y)


def draw_animal(label):
    """Draw a simple animal face. label 0 = cat (pointed ears), 1 = dog (floppy ears + snout)."""
    bg = tuple(int(v) for v in rng.integers(150, 255, 3))
    img = Image.new("RGB", IMG_SIZE, bg)
    d = ImageDraw.Draw(img)
    cx, cy = 32 + int(rng.integers(-4, 5)), 36 + int(rng.integers(-4, 5))
    r = int(rng.integers(14, 19))
    base = [int(v) for v in rng.integers(60, 200, 3)]
    color = tuple(base)
    dark = tuple(max(0, v - 50) for v in base)
    light = tuple(min(255, v + 50) for v in base)

    if label == 0:   # CAT: two triangular ears
        d.polygon([(cx - r, cy - r + 8), (cx - r + 4, cy - r - 10), (cx - r + 13, cy - r + 2)], fill=color)
        d.polygon([(cx + r, cy - r + 8), (cx + r - 4, cy - r - 10), (cx + r - 13, cy - r + 2)], fill=color)
        d.ellipse([cx - r, cy - r + 4, cx + r, cy + r + 2], fill=color)
        for s in (-1, 1):   # whiskers
            d.line([(cx + s * 6, cy + 6), (cx + s * 20, cy + 3)], fill=dark, width=1)
            d.line([(cx + s * 6, cy + 8), (cx + s * 20, cy + 10)], fill=dark, width=1)
    else:            # DOG: floppy ears hanging down + snout
        d.ellipse([cx - r - 6, cy - r + 4, cx - r + 6, cy + 10], fill=dark)
        d.ellipse([cx + r - 6, cy - r + 4, cx + r + 6, cy + 10], fill=dark)
        d.ellipse([cx - r, cy - r + 4, cx + r, cy + r + 2], fill=color)
        d.ellipse([cx - 8, cy + 1, cx + 8, cy + 14], fill=light)

    # eyes and nose (same for both)
    d.ellipse([cx - 8, cy - 3, cx - 4, cy + 1], fill=(20, 20, 20))
    d.ellipse([cx + 4, cy - 3, cx + 8, cy + 1], fill=(20, 20, 20))
    d.ellipse([cx - 2, cy + 5, cx + 2, cy + 8], fill=(40, 20, 20))

    arr = np.array(img, dtype="float32")
    arr += rng.normal(0, 22, arr.shape)          # noise makes the task harder
    return np.clip(arr, 0, 255).astype("uint8")


def make_synthetic(n_per_class):
    X, y = [], []
    for label in (0, 1):
        for _ in range(n_per_class):
            X.append(draw_animal(label))
            y.append(label)
    return np.array(X, dtype="uint8"), np.array(y)


if os.path.isdir(REAL_ROOT):
    print(">> Using REAL cat/dog photos from", REAL_ROOT)
    X_all, y_all = load_real("train", TRAIN_PER_CLASS)
    X_test, y_test = load_real("validation", TEST_PER_CLASS)
else:
    print(">> No real photos found -> generating cat/dog images automatically (no download)")
    X_all, y_all = make_synthetic(TRAIN_PER_CLASS)
    X_test, y_test = make_synthetic(TEST_PER_CLASS)

print("Train+Val images:", X_all.shape, "| Test images:", X_test.shape)

# 2. Split (stratified, like your earlier project)
X_train, X_val, y_train, y_val = train_test_split(
    X_all, y_all, test_size=0.2, random_state=42, stratify=y_all)
print("Train:", len(X_train), "| Validation:", len(X_val), "| Test:", len(X_test))

# 3. CNN: 3 convolutional layers (32 -> 64 -> 128 filters)
model = models.Sequential([
    layers.Input(shape=IMG_SIZE + (3,)),
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.Rescaling(1.0 / 255),

    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),    # Conv layer 1
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),    # Conv layer 2
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu", padding="same"),   # Conv layer 3
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(1, activation="sigmoid"),
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss="binary_crossentropy", metrics=["accuracy"])
model.summary()

early_stop = callbacks.EarlyStopping(monitor="val_loss", patience=8,
                                     restore_best_weights=True)
history = model.fit(X_train, y_train, validation_data=(X_val, y_val),
                    epochs=EPOCHS, batch_size=32, callbacks=[early_stop])
print("Stopped after", len(history.history["loss"]), "epochs")

# 4. Evaluate
y_prob = model.predict(X_test).ravel()
y_pred = (y_prob >= 0.5).astype(int)

print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall   :", round(recall_score(y_test, y_pred), 4))
print("F1-score :", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob), 4))
print(classification_report(y_test, y_pred, target_names=["Cat", "Dog"]))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

# 5. Training curves
plt.figure(figsize=(11, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Train")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("Accuracy"); plt.xlabel("Epoch"); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Train")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("Loss"); plt.xlabel("Epoch"); plt.legend()
plt.tight_layout()
plt.show()

# 6. Sample predictions
idx = np.random.choice(len(X_test), 10, replace=False)
plt.figure(figsize=(12, 5))
for k, i in enumerate(idx):
    plt.subplot(2, 5, k + 1)
    plt.imshow(X_test[i])
    plt.title(f"True: {['Cat','Dog'][y_test[i]]}\nPred: {['Cat','Dog'][y_pred[i]]}", fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()

model.save("cnn_dog_cat.keras")
```
